### Clinical Data :) Data Cleaning

```python
import pandas as pd
```

```python
#File lai Load garne
import os 

file_name = "clinical_data.csv"

BASE_DIR = os.path.dirname(os.getcwd())
file_path = os.path.join(BASE_DIR,"dataset",file_name)
```

```python
#dataframe banaune

df = pd.read_csv(file_path)
```

```python
#Data herne

# df.head() #top ma bhaeko 5 ota Rows dincha
# df.info()
# df.describe()


df.head(10)
```

```python
#Check Missing Values

df.isnull().sum()
```

```python
#duplicate value check garne

df.duplicated().sum()

#eddi cha bhane remove garne : df.drop_duplicates(inplace=True)


#yo dataset ma missing value chaina
#and duplicate value chaina
```

```python

#-> Blood Pressure Column ma esto value cha : 165/93
#jaslai Use garna Difficult huncha as Machine le Bujna sakdaina
# Machine le Number matra bujcha 
# so Seperate garne i.e Upper value lai seperate column banaune & lower value lai seperate column


type(df["Blood Pressure"])
df["Blood Pressure"][:5]
df["Blood Pressure"].str #string ma convert garcha

#seperate colume banauna lai
df[["systolic", "diastolic"]] = df["Blood Pressure"].str.split("/",expand=True)
```

```python
df.head()
```

```python
#BP ko Upper value and Lower Value lai Integer ma convert garne Float ma cha bhane 

df["systolic"] = df["systolic"].astype(int)
df["diastolic"] = df["diastolic"].astype(int)
```

```python
#Now Blood Pressure Column lai Remove garne

df.drop("Blood Pressure", axis=1, inplace=True)
```

```python
#Outlier Check garne
#Outlier bhaneko Different Value i.e Abnormal value haru

df.describe()
```

```python
#Column Name change garne -> Underscore naming is better for Machine Learning

df.rename(columns={
    "Patient ID": "patient_id",
    "Heart Rate": "heart_rate",
    "Family History": "family_history",
    "Triglycerides": "triglycerides"
}, inplace=True)
```

```python
#Feature Engineering garne

def bmi_category(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25:
        return "Normal"
    elif bmi < 30:
        return "Overweight"
    else:
        return "Obese"

df["BMI_Category"] = df["BMI"].apply(bmi_category)
```

```python
def bp_category(sys_bp, dia_bp):
    if sys_bp < 120 and dia_bp < 80:
        return "Normal"
    elif sys_bp < 130 and dia_bp < 80:
        return "Elevated"
    elif sys_bp < 140 or dia_bp < 90:
        return "High_Stage1"
    else:
        return "High_Stage2"

df["BP_Category"] = df.apply(
    lambda row: bp_category(row["systolic"], row["diastolic"]),
    axis=1
)
```

```python
def hr_zone(hr):
    if hr < 60:
        return "Low"
    elif hr <= 100:
        return "Normal"
    else:
        return "High"

df["HR_Zone"] = df["heart_rate"].apply(hr_zone)
```

```python
import numpy as np 

df["Cholesterol_Risk"] = pd.cut(
    df["Cholesterol"],
    bins=[0, 200, 239, np.inf],
    labels=["Desirable", "Borderline", "High"]
)
```

```python
df["Diabetes_Label"] = df["Diabetes"].map({0: "No", 1: "Yes"})
df["Family_History_Label"] = df["family_history"].map({0: "No", 1: "Yes"})
```
